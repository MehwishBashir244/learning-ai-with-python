```python
# Dictionary
# A dictionary is a Python data type that stores data in key-value pairs.
student = {
    "name": "Ali",
    "age": 20,
    "course": "Python"
}

print(student)
print(type(student))
print(student["name"])
print(student.values())
print(student.keys())
```

```python
#nested dict
student = {
    "name" : "sana",
    "subj" : {
        "phy" : 97,
        "bio" : 99,
    }
}
print(student)
print(student["subj"])
print(student["subj"]["bio"])
```

```python
#Create a nested dictionary for 2 students
students = {
    "student1": {"name": "Alice", "age": 16, "grade": "B"},
    "student2": {"name": "Bob", "age": 17, "grade": "A"}
}

# Access the "grade" of student2
student2_grade = students["student2"]["grade"]
print("Student2's grade:", student2_grade)

# Add a new subject to student1
students["student1"]["subject"] = "Math"
print("Updated students dictionary:", students)
```

```python
#dict methods
dic = {
    "key" : "value",
    "name" : "mehwish",
    "num" : 22,
}
print(dic.keys())
print(list(dic.keys()))
print(len(dic))
print(len(list(dic.keys())))

print(dic.values())
print(list(dic.values()))
print(len(dic))
print(len(list(dic.values())))

print(dic.items())
print(list(dic.items()))

pairs = list(dic.items())
print(pairs[0])

```

```python
# give same value
print(dic.get("name"))
print(dic["name"])

# 2 mehods of update
dic.update({"city" : "Pakpattan"})
print(dic)

new_dict = {"name" : "Bashir", "age" : 22}
dic.update(new_dict)
print(dic)
```

```python
# Merge two dict
student1 = {"name": "Bob", "age": 16}
student2 = {"grade": "A", "school": "XYZ High"}
student1.update(student2)
print(student1)
```

```python
# Remove the last inserted item
my_dict = {"name": "Alice", "age": 25, "city": "New York"}
last_item = my_dict.popitem()  
print("Removed item:", last_item)
print("After removing last item:", my_dict)
```

```python
#Clear all items
my_dict.clear() 
print("After clearing:", my_dict)
```

```python
#Find the key with the maximum value in a dictionary
scores = {
    "Alice": 88, 
    "Bob": 92, 
    "Charlie": 85}
print(max(scores, key=scores.get))
```
