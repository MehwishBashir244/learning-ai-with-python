```python
# list
# list is an ordered, mutable (changeable) collection of items.

marks = [22, 33, 44, 55, 66]
print(marks)
print(type(marks))
print(len(marks))

stu = ["ali", 22, 30.0]
stu[0]="sana"
print(stu)

```

```python
#lists slicing

marks = [22, 33, 44, 55, 66]
print(marks[1:3])
print(marks[:3])
print(marks[1:])
print(marks[-2:-1])
```

```python
#lists methods
lis = [1, 4, 6, 3, 2, 5]
lis.append(7)
print(lis)

# sort in ascending order
lis = [1, 4, 6, 3, 2, 5]
print(lis.sort())
print(lis)

# sort in descending order
lis = [1, 4, 6, 3, 2, 5]
print(lis.sort(reverse=True))
print(lis)

lis = [1, 4, 6, 3, 2, 5]
lis.reverse()
print(lis)

lis = [1, 4, 6, 3, 2, 5]
lis.insert(0,0)
print(lis)

lis = [1, 4, 6, 3, 2, 5]
lis.remove(6)
print(lis)

lis = [1, 4, 6, 3, 2, 5]
lis.pop(0)
print(lis)
```

```python
# list half swapping

l = [1, 2, 3, 4, 5, 6]
# Find the middle index
middle = len(l) // 2
swapped_list = l[middle:] + l[:middle]
print("Original list:", l)
print("Swapped list:", swapped_list)


# Display the fourth element of the following list using print()

prime_numbers = [2, 3, 5, 7, 11]
print(prime_numbers[3])


# Add one name to friends_list using append

friends_list = ["Tommy", "Isabel", "Daniel"]
friends_list.append("Otto")

print(friends_list)
```
