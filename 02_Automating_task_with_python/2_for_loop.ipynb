```python
#for loop
# A for loop is used to repeat a block of code for each item in a sequence

nums = [1, 6, 9, 12, 45]
for val in nums:
    print(val)

```

```python
str = "mehwish"
for char in str:
    print(char)
for char in str:
    if(char == "w"):
        print("w found")
        break
    print(char)
else:
    print("end")
```

```python
#calculate the sum of numbers in a list
nums = [2, 5, 7, 10] 
sum = 0
for num in nums:
    sum += num  
print("Sum of elements:", sum)
```

```python
#Print each word in a sentence
sentence = input("Enter a sentence: ")
for word in sentence.split():
    print(word)
```

```python
#Multiply all numbers in a list
l = [2, 3, 4]  
product = 1         
for num in l:
    product *= num   
print("Product of all numbers:", product)
```

```python
#Print letters in reverse of a string (without using range)
text = input("Enter a string: ")
for char in text[::-1]:
    print(char)
```

```python
#Print a triangle pattern using a list of strings
rows = ["*", "**", "***", "****"]
for row in rows:
    print(row)
```
