```python
# ---f-string in python---
# An f-string is used to easily put variables or expressions inside a string.

name = "Mehwish"
course = "AI"
print(f"My name is {name} and i am learning {course}.")


days = 365
print(f"There are {days} days in a year.")


a = 10
b = 5
print(f"The sum is {a + b}")

```
