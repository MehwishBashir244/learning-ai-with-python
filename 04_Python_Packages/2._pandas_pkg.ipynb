```python
#dropna(returns a new DataFrame, and will not change the original) to remove rows of empty cell
df = pd.read_csv('da.csv')
n = df.dropna()
print(n.to_string())
```

```python
# inplace = True (If you want to change the original DataFrame), it will remove all rows containing NULL
df.dropna(inplace = True)
print(df.to_string())
```

```python
import pandas as pd

#fillna (allows us to replace empty cells with a value)
df = pd.read_csv('da.csv')
df.fillna(120, inplace = True)
print(df.to_string())
```
