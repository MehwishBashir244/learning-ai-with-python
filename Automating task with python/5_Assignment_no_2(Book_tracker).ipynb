```python
# Exercise: The Bookworm's Inventory 📖
# Imagine you're a librarian, and a request just came in for the classic novel "To Kill a Mockingbird". 🐦 Before you can tell the eager reader if it's available, you need to check your records.

# You first need to represent a book's information using a Python dictionary. This dictionary will store the following details about the book:

# o title (string): The title of the book.
# o author (string): The author of the book.
# o on_shelf (boolean): Whether the book is currently on the shelf (True) or checked out (False).
# o borrower (string): The name of the person who has borrowed the book (if it's checked out).
# o overdue (boolean): Whether the book is overdue (True) or not (False).
# o on_hold (boolean): Whether the book is on hold for another patron (True) or not (False).
# o Your Task: Initialize the dictionary book with the following key/value pairs:

# o title: To Kill a Mockingbird
# o author: Harper Lee
# o on_shelf: False
# o borrower: Arthur Dent
# o overdue: True
# o on_hold: False
```

```python
# How do you create a dictionary for the book "To Kill a Mockingbird" with its title, author, shelf status, borrower, overdue status, and hold status?

book = {
    "title": "To Kill a Mockingbird",
    "author": "Harper Lee",
    "on_shelf": False,
    "borrower": "Arthur Dent",
    "overdue": True,
    "on_hold": False
}
```

```python
# Write an if/else statement to check whether the book is available to borrow.If the book is on_shelf (meaning on_shelf is True) and it's not on_hold (meaning on_hold is False), print: "Book is available to be borrowed".Else, print: "Book is not available to be borrowed".

if book["on_shelf"] == True and book["on_hold"] == False:
    print("Book is available to be borrowed")
else:
    print("Book is not available to be borrowed")
```

```python
# Write an if/else statement to check whether the book is overdue. If it is overdue, display the borrower's name. Otherwise, put the book on hold.

if book["overdue"] == True:
    print(f"Book is overdue - Contact {book['borrower']} to return it")
else:
    book["on_hold"] = True
    print("Book has been put on hold")
```
