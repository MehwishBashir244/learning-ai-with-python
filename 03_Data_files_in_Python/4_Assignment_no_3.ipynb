```python
# Define a Python function called read_article that takes text_file as a parameter, reads the contents of the file, and returns the contents.

def read_article(text_file):

    f = open(text_file, "r")
    contents = f.read()
    f.close()
    return contents
```

```python
news_article = read_article("news_article.txt")
print(news_article)
```
