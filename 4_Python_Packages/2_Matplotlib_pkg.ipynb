```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
marks = [50, 55, 65, 70, 80]

plt.plot(hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
marks = [50, 55, 65, 70, 80]

plt.scatter(hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "Math", "English", "Physics"]
marks = [85, 70, 80, 75]

plt.bar(subjects, marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```
