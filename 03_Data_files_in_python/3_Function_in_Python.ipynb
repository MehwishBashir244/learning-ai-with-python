```python
#functions
# function is a reusable block of code that performs a specific task.

def sum(a,b):
    sum = a + b
    print(sum)
    return sum

sum(10, 11)
sum(1, 1)

def cal_sum(a,b): # parameters
    return a+b

sum = cal_sum(1, 2) # functin call
print(sum)


def print_hi():
    print("hi")
print_hi()
```

```python
# calculate average

def cal_avg(a, b ,c):
    sum = a + b + c
    avg = sum / 3
    print("Average:",avg)

cal_avg( 22,45,67)
```

```python
# for multiplication

def cal_product(a, b):
    print("product is",a*b)

cal_product(22,45)
```

```python
# for length of list

lis = ["apple","banana", "grapes"]
def print_len(list):
    print(len(list))
print_len(lis)
```

```python
# to cal factorial of a number

def cal_fact(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    print("factorial of the number is",fact)

cal_fact(4)
```

```python
# to convert money

def converter(usd_val):
    inr_val = usd_val * 83
    print(usd_val, "USD:", "=", inr_val,"INR")
converter(2)
```

```python
#Even or Odd

def is_even_or_odd(n):
    if n % 2 == 0:
        print("The number is Even")
    else:
        print("The number is Odd") 

is_even_or_odd(43)  
```

```python
#Maximum of Two Numbers

def maximum(a, b):
    if a > b:
        print("max", a)
        return a
    else:
        print("max", b)
        return b 
    
maximum(2, 5)
```

```python
# 6. Positive, Negative, or Zero

def check_number(n):
    if n > 0:
        return "The number is Positive"
    elif n < 0:
        return "The number is Negative"
    else:
        return "The number is Zero"
    
n =check_number(2)
print(n)
```

```python
#Convert Celsius to Fahrenheit

def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

n = celsius_to_fahrenheit(2)
print("Temperature in fahrenheit is", n)
```

```python
#Print Numbers from 1 to N

def print_numbers(n):
    for i in range(1, n+1):
        print(i, end=" ")
    print()

n = print_numbers(10)
print(n)
```
