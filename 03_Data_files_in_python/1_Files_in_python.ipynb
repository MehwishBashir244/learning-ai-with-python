```python
#for reading line by line
f = open("ai.txt", "r")
line1 = f.readline()
print(line1)
line4 = f.readline()
print(line4)
f.close()
```

```python
# writing to file

f = open("ai.txt", "w") # w remove all data and add new
f.write("AI is fun if you understand it")
f.close()

f = open("ai.txt", "a") # add at last of previous data
f.write("AI is fun if you understand it")
f.close()
```

```python
# to count words, lines or char

with open("ai.txt", "r") as file:
        lines = file.readlines()
        
total_lines = len(lines)
total_words = sum(len(line.split()) for line in lines)
total_chars = sum(len(line) for line in lines)
    
print(f"Total lines: {total_lines}")
print(f"Total words: {total_words}")
print(f"Total characters: {total_chars}")
```
